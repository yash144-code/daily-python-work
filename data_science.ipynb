### More Complex `for` loop Example

This example uses a `for` loop to process a list of student dictionaries. For each student, it checks their `score`. If the score is 60 or higher, it prints a 'Pass' message; otherwise, it prints 'Fail'.

```python
students = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 55},
    {"name": "Charlie", "score": 70},
    {"name": "Diana", "score": 40}
]

print("--- Student Performance ---")
for student in students:
  name = student["name"]
  score = student["score"]

  if score >= 60:
    print(f"{name} passed with a score of {score}.")
  else:
    print(f"{name} failed with a score of {score}.")
print("---------------------------")
```

```python
import numpy as np
import matplotlib.pyplot as plt
hours = np.array([1, 2, 3, 4, 5])
marks = np.array([35, 45, 55, 65, 75])
m, c = np.polyfit(hours, marks, 1)
plt.scatter(hours, marks, label="Data points")

# trend line values
y_line = m * hours + c

plt.plot(hours, y_line, color="red", label="Trend line")

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks with Trend Line")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

ad_spend = np.array([5, 8, 10, 4, 7, 12, 15, 9, 6, 11, 10, 16, 3, 14, 8, 11, 15, 6, 13, 17])  # in lakhs
sales = np.array([20, 30, 38, 18, 28, 45, 55, 34, 24, 42, 40, 60, 15, 52, 32, 44, 56, 25, 48, 65])  # in lakhs

m,c = np.polyfit(ad_spend,sales,1)

plt.scatter(ad_spend,sales,marker = "^",label = "data points")

y_line = m * ad_spend + c

plt.plot(ad_spend,y_line,color = "r",linestyle = "-." ,label = "Trend Line")

plt.xlabel('ad_spend')
plt.ylabel('sales')
plt.title('sales by ads')
plt.legend()
plt.grid(True)
plt.show()
```

When `sparse_output` is set to `True` (which is the default in older versions of scikit-learn, though `sparse_output` is now the preferred name over `sparse`), the encoder returns a SciPy sparse matrix. Sparse matrices are efficient for storing data where most values are zero, as they only store the non-zero elements, saving memory and computation time.

When `sparse_output` is set to `False` (as in your previous code), the encoder returns a dense NumPy array. This is a regular array where all elements, including zeros, are explicitly stored. This format is often easier to work with directly, especially for smaller datasets or if subsequent operations require dense arrays.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

data = {
    "color": ['Red', 'Blue', 'Green', 'Red']
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("\n--------------------------------------------------\n")

# OneHotEncoder with sparse_output=True (or sparse=True in older versions)
encoder_sparse = OneHotEncoder(sparse_output=True)
encoded_sparse = encoder_sparse.fit_transform(df[['color']])

print("Output with sparse_output=True (SciPy sparse matrix):")
print(encoded_sparse)
print("Type:", type(encoded_sparse))
print("\n--------------------------------------------------\n")

# OneHotEncoder with sparse_output=False
encoder_dense = OneHotEncoder(sparse_output=False)
encoded_dense = encoder_dense.fit_transform(df[['color']])

print("Output with sparse_output=False (NumPy dense array):")
print(encoded_dense)
print("Type:", type(encoded_dense))

# You can convert a sparse matrix to a dense array using .toarray()
print("\nSparse matrix converted to dense array:")
print(encoded_sparse.toarray())
```

```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# dataset
data = {
    "Age": [25, None, 30, 35],
    "City": ["Delhi", "Mumbai", "Delhi", "Chennai"]
}

df = pd.DataFrame(data)

# numerical pipeline
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean"))
])

# categorical pipeline
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder())
])

# ColumnTransformer
preprocessor = ColumnTransformer([
    ("num", num_pipeline, ["Age"]),
    ("cat", cat_pipeline, ["City"])
])

# apply transformation
result = preprocessor.fit_transform(df)

print(result)
```

```python
import pandas as pd
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.compose import ColumnTransformer

# dataset
data = {
    "Age": [22, 25, np.nan, 30, 28],
    "Salary": [50000, 60000, 55000, np.nan, 52000],
    "City": ["Delhi", "Mumbai", "Delhi", "Chennai", np.nan]
}

df = pd.DataFrame(data)

X = df
y = [0, 1, 0, 1, 0]

# numerical pipelines (separate for each column)
age_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean"))
])

salary_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

# categorical pipeline
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# ColumnTransformer
preprocessor = ColumnTransformer([
    ("age", age_pipeline, ["Age"]),
    ("salary", salary_pipeline, ["Salary"]),
    ("city", cat_pipeline, ["City"])
])

# full pipeline
pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LogisticRegression())
])

# train
pipeline.fit(X, y)

# prediction
new_data = pd.DataFrame({
    "Age": [27],
    "Salary": [58000],
    "City": ["Delhi"]
})

print(pipeline.predict(new_data))
```

```python
import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

data = {
    "Age": [22, np.nan, 35, 40, 28],
    "Experience": [1, 3, np.nan, 10, 5],
    "Department": ["HR", "IT", "HR", np.nan, "Finance"],
    "Gender": ["Male", "Female", np.nan, "Female", "Male"]
}

df = pd.DataFrame(data)

X = df
y = [0, 1, 0, 1, 0]

#num pipeline
Age_pipeline = Pipeline([("imputer",SimpleImputer(strategy="mean"))])

Exp_pipeline = Pipeline([("imputer",SimpleImputer(strategy = "median"))])

#categorical pipeline
Dep_pipeline = Pipeline([("imputer",SimpleImputer(strategy = "most_frequent")),("encoder",OneHotEncoder(handle_unknown= "ignore"))])
gen_pipeline = Pipeline([("imputer",SimpleImputer(strategy = "most_frequent")),("encoder",OneHotEncoder(handle_unknown= "ignore"))])

#column trans
preprocessor = ColumnTransformer([
    ("Age",Age_pipeline,["Age"]),
    ("experience",Exp_pipeline,["Experience"]),
    ("Department",Dep_pipeline,["Department"]),
    ("Gender",gen_pipeline,["Gender"])
])

pipeline = Pipeline([
    ("preprocessing",preprocessor),
    ("model",LogisticRegression())
])
pipeline.fit(X, y)

new_data = pd.DataFrame({
    "Age": [30],
    "Experience": [4],
    "Department": ["IT"],
    "Gender": ["Male"]
})
print(pipeline.predict(new_data))
```

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import FunctionTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler




data = {
    "Age": [20, 25, np.nan, 30, 35],
    "Salary": [20000, 50000, 80000, np.nan, 120000],
    "City": ["Delhi", "Mumbai", "Delhi", "Chennai", np.nan]
}

df = pd.DataFrame(data)

X = df
y = [0, 1, 0, 1, 1]

#num pipeline
Age_pipeline = Pipeline([("imputer",SimpleImputer(strategy = "mean")),("scaler",StandardScaler())])

Sal_pipeline = Pipeline([("imputer",SimpleImputer(strategy = "median")),("log", FunctionTransformer(np.log1p)),("scaler",StandardScaler())])

#cat pieline
# Removed FunctionTransformer(np.log1p) as it's not suitable for categorical data
City_pipeline = Pipeline([("imputer",SimpleImputer(strategy = "most_frequent")),("encoder",OneHotEncoder(handle_unknown="ignore"))])

preprocessor = ColumnTransformer([
    ("num",Age_pipeline,["Age"]),
    ("sal",Sal_pipeline,["Salary"]),
    ("cat",City_pipeline,['City'])
])

pipeline = Pipeline([
    ("preprocessing",preprocessor),
    ("model",LogisticRegression())
])
pipeline.fit(X,y)

new_data = pd.DataFrame({
    "Age": [27],
    "Salary": [60000],
    "City": ["Delhi"]
})

print(pipeline.predict(new_data))
```

```python
class BankAccount:

    def __init__(self, holder, balance):

        self.holder = holder
        self.balance = balance


    def show_balance(self):

        print("Balance:", self.balance)


    def deposit(self, amount):

        self.balance += amount

        print("Amount Deposited")


    def withdraw(self, amount):

        if amount > self.balance:

            print("Insufficient Funds")

        else:

            self.balance -= amount

            print("Withdrawal Successful")


acc1 = BankAccount("Yash",5000)

acc1.show_balance()

acc1.deposit(2000)

acc1.show_balance()

acc1.withdraw(1000)

acc1.show_balance()
```

```python
import numpy as np

sales = np.array([1200, 1500, 1800, 1100, 2000])
days = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5"]

# Total sales
total_sales = np.sum(sales)
print(f"Total sales: {total_sales}")

# Average sales
average_sales = np.mean(sales)
print(f"Average sales: {average_sales}")

# Best sales day using argmax()
best_sales_day_index = np.argmax(sales)
best_sales_value = sales[best_sales_day_index]
best_day = days[best_sales_day_index]
print(f"Best sales day: {best_day} with sales of {best_sales_value}")
```

Data Normalization Practice

Given:

arr = np.array([10,20,30,40,50])

Normalize using:

(arr - arr.mean()) / arr.std()

🔥 VERY important ML preprocessing concept.

```python
import numpy as np

arr = np.array([10,20,30,40,50])

mean = np.mean(arr)
st_d = np.std(arr)

normalize = (arr - mean) / st_d
print(normalize)

print(normalize.mean())
print(normalize.std())
```

```python
import pandas as pd

data = {
    "Name": ["A","B","C","D","E"],
    "Marks": [90,75,88,67,95],
    "Age": [21,22,20,23,21],
    "City": ["Hyd","Guntur","Vij","Hyd","Warangal"]
}

df = pd.DataFrame(data)
print(df)
print("\n")

highest_marks = df.loc[:,"Marks"].idxmax()
print(df.loc[highest_marks])
print("\n")

lowest_marks_student_city= df.loc[df["Marks"].idxmin(),"City"]
print(f"city of student with lowest marks :{lowest_marks_student_city}")
print("\n")

youngest_student = df.loc[df["Age"].idxmin()]
print(youngest_student)
print("\n")

print(df.loc[df["Age"].idxmax(),"Name"])
```

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["A","B","C","D","E","F"],
    "Marks": [90,75,88,67,95,72],
    "Age": [21,22,20,23,21,24],
    "City": ["Hyd","Guntur","Hyd","Vij","Warangal","Hyd"]
}

df = pd.DataFrame(data)

conditions = [
    df["Marks"] > 80,
    df["Marks"].between(70, 80)
]
choices = ["Excellent", "Good"]

df["Status"] = np.select(conditions, choices, default="Average")

print(df)
```

3️⃣5️⃣ Which category generated:

highest revenue?

3️⃣6️⃣ Which gender spends more on average?

3️⃣7️⃣ Which age group appears most frequently?   

3️⃣8️⃣ Find top 5 biggest transactions.

3️⃣9️⃣ Which category has highest average quantity sold?

4️⃣0️⃣ Which month had highest sales?


```python
import pandas as pd

df = pd.read_csv("Sales Dataset.csv")
df["Date"] = pd.to_datetime(df["Date"])

sales_category = pd.pivot_table(df, values = "Total Amount", index  = "Product Category", aggfunc= "sum")

Category = sales_category["Total Amount"].idxmax()

highest_rev =  sales_category["Total Amount"].max()

print(f"The category that gave highest revenve is : {Category} with sales of {highest_rev}")
print("\n")

mean_gender = pd.pivot_table(df, values = "Total Amount", index  = "Gender", aggfunc= "mean")
Gender = mean_gender["Total Amount"].idxmax()
highest_avg_spend = mean_gender["Total Amount"].max()
print(f"the gender that sends more on average is : {Gender} with average spend of {highest_avg_spend}")
print("\n")

print(f"the age group that appears most frequent is: {df["Age"].value_counts().idxmax()}")
print("\n")

print(df.nlargest(5,"Total Amount"))
print("\n")

average_quantity = pd.pivot_table(df,values = "Quantity", index = "Product Category", aggfunc = "mean")
highest_category = average_quantity["Quantity"].idxmax()
highest_avg_quant = average_quantity["Quantity"].max()

print(f"the product category with highest average quantity is {highest_category} with an average of {highest_avg_quant}")
print("\n")

df["Month"] = df["Date"].dt.month_name()
monthly_sales = pd.pivot_table(df, values = "Total Amount" ,index = "Month" , aggfunc = "sum")
highest_sold_month = monthly_sales["Total Amount"].idxmax()
highest_sales = monthly_sales["Total Amount"].max()
print(f"Month with highest sales is {highest_sold_month} with sales of {highest_sales}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Sales Dataset.csv")

df["Date"] = pd.to_datetime(df["Date"])
df["Month"] = df["Date"].dt.month_name()

month_order = ["January","February","March","April","May","June","July","August","September","October","November","December"]
df["Month"] = pd.Categorical(df["Month"],categories= month_order,ordered = True)

plt.figure(figsize=(10,9))

plt.subplot(2,2,1)
monthly_sales = df.groupby("Month")["Total Amount"].sum()
plt.plot(monthly_sales.index,monthly_sales.values,color ="black",marker ="o")
plt.title("monthly sales chart")

plt.grid(True)
plt.xticks(rotation=90)

plt.subplot(2,2,2)
category_sales = df.groupby("Product Category")["Total Amount"].sum()
plt.bar(category_sales.index,category_sales.values,color = "red",edgecolor = "black")
plt.grid(axis ="y")
plt.title("category wise sales chart")

plt.subplot(2,2,3)
gender_counts = df['Gender'].value_counts()
plt.pie(gender_counts, labels = gender_counts.index, colors= ["olive","blue"],startangle = 90,autopct="%1.1f%%")
plt.title("Gender Distribution")

plt.subplot(2,2,4)
plt.hist(df["Age"],color = "red",edgecolor = "black",bins = 10)
plt.title("Age Distribution")


plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("Sales Dataset.csv")

df["Age Group"] = df["Age"].apply(lambda x: "Adult" if x < 40 else "Senior")

adult_sales = (df[df["Age Group"] == "Adult"].groupby("Product Category")["Total Amount"].sum())

senior_sales = (df[df["Age Group"] == "Senior"].groupby("Product Category")["Total Amount"].sum())

categories = adult_sales.index

x = np.arange(len(categories))

width = 0.35

plt.figure(figsize=(10,5))

plt.bar(x - width/2,adult_sales.values,width,label="Adult",color="blue")

plt.bar(x + width/2,senior_sales.values,width,label="Senior",color="red")

plt.xticks(x, categories)

plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.title("Adult vs Senior Category-wise Sales")

plt.legend()

plt.grid(axis="y")

plt.show()
```
